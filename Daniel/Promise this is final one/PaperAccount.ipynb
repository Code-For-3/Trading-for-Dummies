### Place Limit Orders

```python
url = "https://demo.trading212.com/api/v0/equity/orders/limit"

payload = {
  "limitPrice": 100.23,
  "quantity": 0.1,
  "ticker": "AAPL_US_EQ",
  "timeValidity": "DAY"
}

headers = {
  "Content-Type": "application/json",
  "Authorization": "YOUR_API_KEY_HERE"
}

response = requests.post(url, json=payload, headers=headers)

data = response.json()
print(data)
```
