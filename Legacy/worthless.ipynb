```python
import vectorbt as vbt
import numpy as np

# Example Close Prices
close = np.array([1, 2, 3, 4, 5], dtype=np.float64)

# Run RSI using vectorbt's TA-Lib wrapper
rsi_ind = vbt.IndicatorFactory.from_talib("RSI").run(close=close, timeperiod=2)

# Access RSI values
print(rsi_ind.real.to_numpy())
```
